```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
# 1. Load Data
df_nasabah = pd.read_csv('nasabah_raw.csv')
df_kredit = pd.read_csv('data_kredit_eksternal.csv')
```

```python
# 2. Pemeriksaan Awal & Hapus Duplikat (Tugas A)
print(f"Dimensi awal: {df_nasabah.shape}")
print(f"Jumlah duplikat: {df_nasabah.duplicated().sum()}")
df_clean = df_nasabah.drop_duplicates().copy()
```

```python
# 3. Penanganan Kesalahan Data (Tugas B)
df_clean['Status_Pernikahan'] = df_clean['Status_Pernikahan'].astype(str).str.strip().str.title()
df_clean.loc[(df_clean['Usia'] < 17) | (df_clean['Usia'] > 100), 'Usia'] = np.nan
```

```python
# 4. Integrasi Data / Left Join (Tugas C)
df_merged = pd.merge(df_clean, df_kredit, on='ID_Pelanggan', how='left')
print(f"Dimensi setelah join: {df_merged.shape}")
```

```python
# 5. Penanganan Missing Values (Tugas D)
for col in ['Usia', 'Pendapatan_Bulanan', 'Lama_Bekerja_Thn', 'Pendapatan_Tahunan_Est']:
    df_merged[col].fillna(df_merged[col].median(), inplace=True)
    df_merged['Jumlah_Kartu_Kredit'].fillna(0, inplace=True)
df_merged['Tingkat_Pendidikan'].fillna(df_merged['Tingkat_Pendidikan'].mode()[0], inplace=True)
df_merged['Pernah_Gagal_Bayar_Lain'].fillna(df_merged['Pernah_Gagal_Bayar_Lain'].mode()[0], inplace=True)
```

```python
# 6. Penanganan Outlier (Tugas E)
Q1, Q3 = df_merged['Pendapatan_Bulanan'].quantile([0.25, 0.75])
IQR = Q3 - Q1
df_merged['Pendapatan_Bulanan'] = df_merged['Pendapatan_Bulanan'].clip(Q1 - 1.5 * IQR, Q3 + 1.5 * IQR)
```

```python
# Simpan hasil akhir bersih ke CSV (Instruksi A.4)
df_merged.to_csv('clean_nasabah_final.csv', index=False)
print("File 'clean_nasabah_final.csv' berhasil disimpan!")
```

```python
# 7. Categorical Encoding (Tugas F)
df_encoded = df_merged.copy()
df_encoded['Tingkat_Pendidikan'] = df_encoded['Tingkat_Pendidikan'].map({'SMA': 1, 'Sarjana': 2, 'Magister': 3})
df_encoded['Pernah_Gagal_Bayar_Lain'] = df_encoded['Pernah_Gagal_Bayar_Lain'].map({'Ya': 1, 'Tidak': 0})
status_dummies = pd.get_dummies(df_encoded['Status_Pernikahan'], prefix='Status_Pernikahan', drop_first=True, dtype=int)
df_encoded = pd.concat([df_encoded, status_dummies], axis=1).drop(columns=['ID_Pelanggan', 'Status_Pernikahan'])
```

```python
# 8. Penskalaan Fitur / Scaling (Tugas G)
X = df_encoded.drop(columns=['Status_Default'])
y = df_encoded['Status_Default']

num_cols = ['Usia', 'Pendapatan_Bulanan', 'Skor_Kredit', 'Lama_Bekerja_Thn', 'Pendapatan_Tahunan_Est', 'Jumlah_Kartu_Kredit']
print("\nStatistik 3 Fitur Sebelum Scaling:")
print(X[['Usia', 'Pendapatan_Bulanan', 'Skor_Kredit']].describe())

scaler = StandardScaler()
X_scaled = X.copy()
X_scaled[num_cols] = scaler.fit_transform(X[num_cols])

print("\nStatistik 3 Fitur Setelah Scaling (Mean ~0, Std ~1):")
print(X_scaled[['Usia', 'Pendapatan_Bulanan', 'Skor_Kredit']].describe())
```
